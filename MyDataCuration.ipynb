# Data Wrangling

Filter data according to trageted years, service and fuel type

```python
df_filtered = df[(df['date'] >= '2000-01') & (df['date'] <= '2023-01') & (df['service'].isin(['self', 'no type'])) & (df['fuel'].isin(['gasoline', 'natural gas']))]
df_filtered.head(25)
```

```python
df_filtered.city.value_counts()
```

```python
df_filtered.fuel.value_counts()
```

```python
df_filtered.describe(include = 'all')
```

Filter out cities with no data on natural gas

```python
df_filtered = df_filtered[~df_filtered['city'].isin(['Calgary', 'Edmonton'])]
df_filtered.head()
```

Feature engineer the fuel type column by combining the fuel and octane columns

```python
df_filtered = df_filtered.copy()
df_filtered['fuel_type'] = df_filtered['fuel'] + "_" + df_filtered['octane']
df_filtered.head()
```

Now check for any null values

```python
df_filtered.isnull().any()
```

Group fuel types in different columns based on cents per litre and octane type

```python
df_pivot = df_filtered.pivot_table(
    index=['date', 'city', 'province'],
    columns='fuel_type',
    values='cents_per_litre'
).reset_index()
df_pivot.columns.name = None
df_pivot.head(25)
```

Rename the pivoted column name

```python
df_pivot.rename(columns={"natural gas_no type": "natural gas"}, inplace=True)
df_pivot.head(25)
```

Convert the date column to datetime format and Feature engineer the year column

```python
df_pivot['date'] = pd.to_datetime(df_pivot['date'], format="%Y-%m")
df_pivot['Year'] =df_pivot['date'].dt.year
df_pivot.head()
```

Feature engineer the month column

```python
df_pivot['month'] = df_pivot['date'].dt.month
df_pivot['month_name'] = df_pivot['date'].dt.month_name()
df_pivot.head()
```

Feature engineer the seasonal patterns column

```python
df_pivot['season'] = df_pivot['month'].map({
    12: 'Winter', 1: 'Winter', 2: 'Winter',
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Fall', 10: 'Fall', 11: 'Fall'
})
df_pivot.tail()
```

Feature engineer the difference between the prices of regular gasoline and natural gas

```python

df_pivot['regular_natgas_diff'] = df_pivot['gasoline_regular'] - df_pivot['natural gas']

```

Feature engineer the difference between the prices of premium gasoline and natural gas

```python
df_pivot['premium_natgas_diff'] = df_pivot['gasoline_premium'] - df_pivot['natural gas']
df_pivot.head()
```

Sort the values by city date and province
 and  
 Calculate the monthly change in price of gasoline and natural gas

```python
df_pivot = df_pivot.sort_values(['city', 'province', 'date'])
df_pivot['regular_monthly_change'] = df_pivot.groupby(['city', 'province'])['gasoline_regular'].diff()
df_pivot['premium_monthly_change'] = df_pivot.groupby(['city', 'province'])['gasoline_premium'].diff()
df_pivot['natgas_monthly_change'] = df_pivot.groupby(['city', 'province'])['natural gas'].diff()
df_pivot.tail()
```
